```python
!pip install Sastrawi

import re
import string
import pandas as pd
import nltk # Added import statement for nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

def cleaningText(text):
    text = re.sub(r"http\S+", ' ', text) # menghapus link
    text = re.sub(r'[0-9]+', ' ', text) # menghapus angka
    text = re.sub(r'[^\w\s]', ' ', text) # menghapus karakter selain huruf dan angka
    text = re.sub(r'\b(\w{4,})nya\b', r'\1 nya', text)
    text = re.sub(r'\b(\w{4,})nyaa\b', r'\1 nyaa', text)

    text = text.replace('\n', ' ') # mengganti baris baru dengan spasi
    text = text.strip(' ') # menghapus karakter spasi dari kiri dan kanan teks
    return text

def casefoldingText(text): # Mengubah semua karakter dalam teks menjadi huruf kecil
    return text.lower()

def tokenizingText(text): # Memecah atau membagi string, teks menjadi daftar token
    return word_tokenize(text)

def filteringText(text): # Menghapus stopwords dalam teks
    listStopwords = set(stopwords.words('indonesian'))
    listStopwords.update(set(stopwords.words('english')))
    listStopwords.update(['iya','na','sih','ku',
                          'ya','loh','kah','woi','woii','woy'])

    # kata yang penting untuk ABSA jangan dihapus meski masuk stopword
    whitelist = {'tempat', 'harga', 'rasa', 'makan', 'layan', 'parkir', 'baik',
                 'toilet', 'mushola', 'antri', 'tunggu', 'bersih', 'kotor', 'lama', 'sangat'}
    listStopwords -= whitelist

    return [w for w in text if w not in listStopwords]

def stemmingText(text): # Mengurangi kata ke bentuk dasarnya yang menghilangkan imbuhan awalan dan akhiran atau ke akar kata
    # Membuat objek stemmer
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()

    # Memecah teks menjadi daftar kata
    words = text.split()

    # Menerapkan stemming pada setiap kata dalam daftar
    stemmed_words = [stemmer.stem(word) for word in words]

    # Menggabungkan kata-kata yang telah distem
    return ' '.join(stemmed_words)


def toSentence(list_words): # Mengubah daftar kata menjadi kalimat
    return ' '.join(word for word in list_words)
```

```python
import pandas as pd
import nltk
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True) # Added to fix LookupError

# --- Re-running previous steps to ensure df_reviews_no_duplicates is defined ---
# Assuming the file is named 'dataset_pizza_sentiment.xlsx' and is in the same directory
try:
    df_reviews = pd.read_excel('dataset_pizza_sentiment.xlsx')
except FileNotFoundError:
    print("Error: The file 'dataset_pizza_sentiment.xlsx' was not found.")
    print("Please make sure the file is in the correct directory or provide the full path.")
    # Exit or handle the error appropriately if the file is critical
    raise # Re-raise the error to stop execution if file is not found
except Exception as e:
    print(f"An error occurred: {e}")
    raise # Re-raise other exceptions

# Remove rows with null values in the 'ulasan' column
df_reviews_cleaned = df_reviews.dropna(subset=['ulasan'])

# Remove duplicate rows based on the 'ulasan' column
df_reviews_no_duplicates = df_reviews_cleaned.drop_duplicates(subset=['ulasan'])
# --- End of re-running previous steps ---

# Create a copy of the DataFrame to store preprocessed text
df_preprocessed = df_reviews_no_duplicates.copy()

# Membersihkan teks dan menyimpannya di kolom 'text_clean'
df_preprocessed['text_clean'] = df_preprocessed['ulasan'].apply(cleaningText)

# Mengubah huruf dalam teks menjadi huruf kecil dan menyimpannya di 'text_casefoldingText'
df_preprocessed['text_casefoldingText'] = df_preprocessed['text_clean'].apply(casefoldingText)

# Mengganti kata-kata slang dengan kata-kata standar dan menyimpannya di 'text_slangwords'
df_preprocessed['text_slangwords'] = df_preprocessed['text_casefoldingText'].apply(fix_slangwords)

# Memecah teks menjadi token (kata-kata) dan menyimpannya di 'text_tokenizingText'
df_preprocessed['text_tokenizingText'] = df_preprocessed['text_slangwords'].apply(tokenizingText)

# Menghapus kata-kata stop (kata-kata umum) dan menyimpannya di 'text_stopword'
df_preprocessed['text_stopword'] = df_preprocessed['text_tokenizingText'].apply(filteringText)

# Menggabungkan token-token menjadi kalimat dan menyimpannya di 'text_akhir'

df_preprocessed['text_tosentence'] = df_preprocessed['text_stopword'].apply(toSentence)
df_preprocessed['text_akhir'] = df_preprocessed['text_tosentence'].apply(stemmingText)


# Display the head of the preprocessed DataFrame to show the results
print(df_preprocessed[['ulasan', 'text_tosentence', 'text_akhir']].head(10))
```
