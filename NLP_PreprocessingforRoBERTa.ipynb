Import Data

```python
import pandas as pd

# Assuming the file is named 'google_reviews_richeese_pizza.xlsx'
# and is in the same directory as this notebook.
try:
    df_reviews = pd.read_excel('google_reviews_richeese_pizza.xlsx')
    print("Excel file loaded successfully!")
    print(df_reviews.head())
except FileNotFoundError:
    print("Error: The file 'google_reviews_richeese_pizza.xlsx' was not found.")
    print("Please make sure the file is in the correct directory or provide the full path.")
except Exception as e:
    print(f"An error occurred: {e}")
```

Null Value Handling

```python
df_reviews.info()
```

```python
# Remove rows with null values in the 'ulasan' column
df_reviews_cleaned = df_reviews.dropna(subset=['ulasan'])

print("DataFrame after dropping null values in 'ulasan' column:")
df_reviews_cleaned.info()
```

Data Duplicate Handling

```python
# Check for duplicate rows in the 'ulasan' column
duplicate_rows = df_reviews_cleaned[df_reviews_cleaned.duplicated(subset=['ulasan'])]
print(f"Number of duplicate rows based on 'ulasan': {len(duplicate_rows)}")
```

```python
# Remove duplicate rows based on the 'ulasan' column
df_reviews_no_duplicates = df_reviews_cleaned.drop_duplicates(subset=['ulasan'])

print("DataFrame after dropping duplicate values in 'ulasan' column:")
df_reviews_no_duplicates.info()
```

```python
df_reviews_no_duplicates.to_excel('dataset_pizza_cleaned.xlsx', index=False)
print("Dataset exported to 'dataset_pizza_cleaned.xlsx' successfully!")
```
